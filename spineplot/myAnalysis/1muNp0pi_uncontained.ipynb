```python
"""
Inspect the structure of one OLD and one NEW flat CAF.

This is discovery-only — no comparisons or histograms yet. We need to see:
  1. Top-level directories/trees in each file
  2. All branches in recTree (or whatever the main TTree is named)
  3. Specifically: branches under sr.mc, sr.dlp, sr.dlp_true, sr.hdr
  4. Any schema differences between OLD and NEW

Output this and we'll plan the comparison based on what's actually there.
"""

import uproot

OLD_PATH = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/NewOldStudies/old_file.root"
NEW_PATH = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/NewOldStudies/new_file.root"


def inspect(path, label):
    print(f"\n{'='*80}")
    print(f"{label}: {path.split('/')[-1]}")
    print('='*80)

    f = uproot.open(path)

    # Top-level directory listing
    print("\n--- Top-level keys ---")
    for k in f.keys(recursive=False):
        obj = f[k]
        print(f"  {k}   ({type(obj).__name__})")

    # Find the main tree
    main_tree = None
    for candidate in ["recTree", "recTree;1", "sr", "events"]:
        if candidate in f:
            main_tree = f[candidate]
            print(f"\n--- Main tree: '{candidate}' ---")
            print(f"  Num entries: {main_tree.num_entries}")
            break

    if main_tree is None:
        # Fall back: print every TTree-like object found
        print("\n  No recTree found. Looking for any TTree...")
        for k in f.keys(recursive=True):
            obj = f[k]
            if hasattr(obj, "num_entries"):
                print(f"  Found tree: {k}  ({obj.num_entries} entries)")
        return

    # Filter branches by topic
    all_branches = list(main_tree.keys())
    print(f"\nTotal branches: {len(all_branches)}")

    # Branches related to mc (truth-level)
    print("\n--- Branches matching 'mc' ---")
    mc_branches = [b for b in all_branches if "mc" in b.lower()]
    for b in mc_branches[:50]:
        try:
            interp = main_tree[b].interpretation
            print(f"  {b:60s}  {interp}")
        except Exception:
            print(f"  {b:60s}  <no interpretation>")
    if len(mc_branches) > 50:
        print(f"  ... and {len(mc_branches) - 50} more 'mc' branches")
    print(f"  TOTAL 'mc' branches: {len(mc_branches)}")

    # Branches related to dlp (SPINE interactions)
    print("\n--- Branches matching 'dlp' ---")
    dlp_branches = [b for b in all_branches if "dlp" in b.lower()]
    for b in dlp_branches[:30]:
        try:
            interp = main_tree[b].interpretation
            print(f"  {b:60s}  {interp}")
        except Exception:
            print(f"  {b:60s}  <no interpretation>")
    if len(dlp_branches) > 30:
        print(f"  ... and {len(dlp_branches) - 30} more 'dlp' branches")
    print(f"  TOTAL 'dlp' branches: {len(dlp_branches)}")

    # Branches related to ndlp / nnu / ngenevt etc — the counters we care about
    print("\n--- Counter branches (ndlp, nnu, ngenevt, etc) ---")
    counter_patterns = ["ndlp", "nnu", "ngenevt", "pot", "first_in_subrun"]
    for pattern in counter_patterns:
        matches = [b for b in all_branches if pattern in b.lower()]
        for b in matches[:5]:
            try:
                interp = main_tree[b].interpretation
                print(f"  {b:60s}  {interp}")
            except Exception:
                print(f"  {b:60s}  <no interpretation>")
        if matches:
            print()

    # Branches under hdr
    print("\n--- Branches matching 'hdr' ---")
    hdr_branches = [b for b in all_branches if "hdr" in b.lower()]
    for b in hdr_branches[:40]:
        try:
            interp = main_tree[b].interpretation
            print(f"  {b:60s}  {interp}")
        except Exception:
            print(f"  {b:60s}  <no interpretation>")
    print(f"  TOTAL 'hdr' branches: {len(hdr_branches)}")


for path, label in [(OLD_PATH, "OLD"), (NEW_PATH, "NEW")]:
    inspect(path, label)

# Compare branch lists between files
print(f"\n\n{'='*80}\nSCHEMA DIFF\n{'='*80}")
f_old = uproot.open(OLD_PATH)
f_new = uproot.open(NEW_PATH)

for tree_name in ["recTree", "recTree;1"]:
    if tree_name in f_old and tree_name in f_new:
        keys_old = set(f_old[tree_name].keys())
        keys_new = set(f_new[tree_name].keys())
        only_old = keys_old - keys_new
        only_new = keys_new - keys_old
        print(f"\nUsing tree '{tree_name}'")
        print(f"  Branches in OLD only: {len(only_old)}")
        for b in sorted(only_old)[:30]:
            print(f"    {b}")
        if len(only_old) > 30:
            print(f"    ... and {len(only_old) - 30} more")
        print(f"  Branches in NEW only: {len(only_new)}")
        for b in sorted(only_new)[:30]:
            print(f"    {b}")
        if len(only_new) > 30:
            print(f"    ... and {len(only_new) - 30} more")
        print(f"  Common branches: {len(keys_old & keys_new)}")
        break
```

```python
"""
Event-level counter comparison: where does the 8.3% deficit appear?

For each flat CAF, sum the following per-event counters:
  rec.hdr.ngenevt       — GENIE-generated events (most upstream)
  rec.mc.nnu            — MCTruth neutrinos (post-GENIE, pre-SPINE)
  rec.ndlp_true         — SPINE truth interactions (post-SPINE truth processing)
  rec.ndlp              — SPINE reco interactions (post-SPINE reco)

Then divide each by total POT in the file (from rec.hdr.pot summed, cross-checked
against the TotalPOT TH1D). Compare OLD vs NEW per POT.

Where the new/old ratio drops from ~1.0 to ~0.917 localizes the stage where
events are being lost.

Additional checks:
  - Distribution of POT per readout event (in case beam structure changed)
  - Distribution of each counter per readout event (in case it's not just a flat shift)
  - is_time_contained flag in NEW sample (newly added in SPINE — possible culprit)
"""

import uproot
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

OLD_PATH = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/NewOldStudies/old_file.root"
NEW_PATH = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/NewOldStudies/new_file.root"

OUTPUT_PDF = "event_level_counter_comparison.pdf"

# Branches to read for the per-event counter analysis
COUNTER_BRANCHES = [
    "rec.hdr.pot",
    "rec.hdr.ngenevt",
    "rec.mc.nnu",
    "rec.ndlp_true",
    "rec.ndlp",
]

# Sanity check branches
HEADER_BRANCHES = [
    "rec.hdr.ismc",
    "rec.hdr.first_in_subrun",
    "rec.hdr.evt",
]


def read_pot_th1d(f):
    """Pull total POT from the TotalPOT TH1D in the file."""
    try:
        h = f["TotalPOT"]
        members = h.all_members
        fTsumw = members.get("fTsumw")
        if fTsumw is not None and fTsumw > 0:
            return float(fTsumw)
        # fall back: try .values().sum()
        vals = h.values()
        return float(np.asarray(vals).sum())
    except Exception as e:
        print(f"  TotalPOT read failed: {e}")
        return None


def read_total_events_th1d(f):
    """Pull total events from the TotalEvents TH1D."""
    try:
        h = f["TotalEvents"]
        members = h.all_members
        fTsumw = members.get("fTsumw")
        if fTsumw is not None and fTsumw > 0:
            return float(fTsumw)
        vals = h.values()
        return float(np.asarray(vals).sum())
    except Exception as e:
        print(f"  TotalEvents read failed: {e}")
        return None


def analyze_file(path, label):
    print(f"\n{'='*80}")
    print(f"{label}: {path.split('/')[-1]}")
    print('='*80)

    f = uproot.open(path)

    # File-level POT and events
    pot_th1d = read_pot_th1d(f)
    nev_th1d = read_total_events_th1d(f)
    print(f"\nFile-level histograms:")
    print(f"  TotalPOT TH1D    = {pot_th1d:.4e}" if pot_th1d else "  TotalPOT not read")
    print(f"  TotalEvents TH1D = {nev_th1d:.4e}" if nev_th1d else "  TotalEvents not read")

    # Read per-event counters
    tree = f["recTree"]
    n_entries = tree.num_entries
    print(f"\nrecTree: {n_entries} entries")

    arrs = tree.arrays(COUNTER_BRANCHES + HEADER_BRANCHES, library="np")

    pot       = arrs["rec.hdr.pot"]
    ngenevt   = arrs["rec.hdr.ngenevt"]
    mc_nnu    = arrs["rec.mc.nnu"]
    ndlp_true = arrs["rec.ndlp_true"]
    ndlp      = arrs["rec.ndlp"]
    ismc      = arrs["rec.hdr.ismc"]
    first_in_subrun = arrs["rec.hdr.first_in_subrun"]

    # Sanity: ismc should be all 1 for MC samples
    print(f"\n  ismc == 1 fraction: {(ismc == 1).mean():.4f}")
    print(f"  first_in_subrun count: {first_in_subrun.sum()} of {n_entries}")

    # POT bookkeeping
    pot_sum_perevent = pot.sum()
    # POT is stored per-event in flat CAFs; but typically only the first event
    # in each subrun has the POT, and others are 0. Check distribution.
    print(f"\n  rec.hdr.pot: nonzero entries = {(pot > 0).sum()} of {n_entries}")
    print(f"  rec.hdr.pot: sum = {pot_sum_perevent:.4e}")
    print(f"  rec.hdr.pot: mean (nonzero) = {pot[pot>0].mean():.4e}" if (pot>0).any() else "  rec.hdr.pot: all zero")
    print(f"  rec.hdr.pot: max  = {pot.max():.4e}")
    if pot_th1d is not None:
        print(f"  Ratio sum/TH1D     = {pot_sum_perevent/pot_th1d:.4f}")

    # Use TH1D POT if available, otherwise the per-event sum
    total_pot = pot_th1d if pot_th1d and pot_th1d > 0 else pot_sum_perevent
    print(f"\n  USING TOTAL POT = {total_pot:.4e}")

    # Sum each counter
    sums = {
        "ngenevt":    int(ngenevt.sum()),
        "mc.nnu":     int(mc_nnu.sum()),
        "ndlp_true":  int(ndlp_true.sum()),
        "ndlp":       int(ndlp.sum()),
    }
    print(f"\n  Sums across all entries:")
    for k, v in sums.items():
        rate = v / total_pot if total_pot else float('nan')
        print(f"    {k:15s} = {v:8d}    rate per POT = {rate:.4e}")

    # Distribution of each counter per readout event
    print(f"\n  Per-event distributions:")
    for name, arr in [("ngenevt", ngenevt), ("mc.nnu", mc_nnu),
                       ("ndlp_true", ndlp_true), ("ndlp", ndlp)]:
        unique, counts = np.unique(arr, return_counts=True)
        dist_str = ", ".join(f"{u}:{c}" for u, c in zip(unique[:8], counts[:8]))
        print(f"    {name:12s} mean={arr.mean():.3f}  max={arr.max()}  dist={dist_str}")

    return {
        "label":      label,
        "total_pot":  total_pot,
        "n_entries":  n_entries,
        "ngenevt":    ngenevt,
        "mc_nnu":     mc_nnu,
        "ndlp_true":  ndlp_true,
        "ndlp":       ndlp,
        "pot_per_ev": pot,
        "sums":       sums,
    }


def make_comparison_table(old, new):
    print(f"\n\n{'='*80}")
    print("DIAGNOSTIC TABLE: where does the deficit appear?")
    print('='*80)
    print(f"\n{'Stage':<15s} {'rate_old/POT':>16s} {'rate_new/POT':>16s} {'new/old':>10s}  Interpretation")
    print("-" * 100)

    interpretations = {
        "ngenevt":   "GENIE-generated events (upstream of SPINE)",
        "mc.nnu":    "MCTruth neutrinos (post-GENIE, pre-SPINE)",
        "ndlp_true": "SPINE truth interactions (post-SPINE truth)",
        "ndlp":      "SPINE reco interactions (post-SPINE reco)",
    }

    pot_old = old["total_pot"]
    pot_new = new["total_pot"]

    for key in ["ngenevt", "mc.nnu", "ndlp_true", "ndlp"]:
        n_old = old["sums"][key]
        n_new = new["sums"][key]
        rate_old = n_old / pot_old
        rate_new = n_new / pot_new
        ratio = rate_new / rate_old if rate_old > 0 else float('nan')
        print(f"{key:<15s} {rate_old:>16.4e} {rate_new:>16.4e} {ratio:>10.4f}  {interpretations[key]}")

    # Add the systematics ROOT result for context
    print(f"{'signal (syst)':<15s} {'1.5942e-16':>16s} {'1.4613e-16':>16s} {'0.9167':>10s}  CC0π truth signal (from systematics ROOT)")

    print(f"\nInterpretation guide:")
    print(f"  • If ratio ≈ 1.000 at top and drops at some row → loss occurs at that row's stage")
    print(f"  • If ratio ≈ 0.917 from the top → deficit predates GENIE (POT bookkeeping?)")
    print(f"  • If ratio ≈ 1.000 all the way down → loss is in the CC0π selection itself (we know it's not)")


def make_plots(old, new, output_pdf):
    pdf = PdfPages(output_pdf)

    pot_old = old["total_pot"]
    pot_new = new["total_pot"]

    # Per-readout-event distributions for each counter, normalized by POT
    for var in ["ngenevt", "mc_nnu", "ndlp_true", "ndlp"]:
        arr_old = old[var]
        arr_new = new[var]

        # Pick bins covering both ranges
        max_val = max(arr_old.max(), arr_new.max())
        bins = np.arange(-0.5, max_val + 1.5, 1.0)

        co, edges = np.histogram(arr_old, bins=bins)
        cn, _     = np.histogram(arr_new, bins=bins)
        ro = co / pot_old
        rn = cn / pot_new

        fig, (ax_top, ax_bot) = plt.subplots(2, 1, figsize=(8, 6),
                                              gridspec_kw={"height_ratios": [3, 1]},
                                              sharex=True)
        centers = 0.5*(edges[:-1] + edges[1:])
        widths = np.diff(edges)
        ax_top.bar(centers, ro, width=widths*0.9, alpha=0.5,
                   label=f"OLD ({pot_old:.2e} POT)", edgecolor="C0")
        ax_top.bar(centers, rn, width=widths*0.9, alpha=0.5,
                   label=f"NEW ({pot_new:.2e} POT)", edgecolor="C1")
        ax_top.set_yscale("log")
        ax_top.set_ylabel("events / POT")
        ax_top.set_title(f"Distribution of {var.replace('_', '.')} per readout event")
        ax_top.legend()

        with np.errstate(divide='ignore', invalid='ignore'):
            ratio = np.where(ro > 0, rn/ro, np.nan)
        ax_bot.axhline(1.0, color='k', linestyle='--', alpha=0.5)
        ax_bot.axhline(0.9167, color='r', linestyle=':', alpha=0.5, label="0.917 (signal-tree deficit)")
        ax_bot.plot(centers, ratio, 'o-')
        ax_bot.set_xlabel(var.replace("_", "."))
        ax_bot.set_ylabel("new/old (per POT)")
        ax_bot.set_ylim(0.5, 1.5)
        ax_bot.legend()
        plt.tight_layout()
        pdf.savefig(fig)
        plt.close(fig)

    # POT per readout event distribution
    fig, ax = plt.subplots(figsize=(8, 5))
    p_old = old["pot_per_ev"]
    p_new = new["pot_per_ev"]
    p_old_nz = p_old[p_old > 0]
    p_new_nz = p_new[p_new > 0]
    if len(p_old_nz) > 0 and len(p_new_nz) > 0:
        bins = np.linspace(0, max(p_old_nz.max(), p_new_nz.max()), 40)
        ax.hist(p_old_nz, bins=bins, alpha=0.5, label=f"OLD (nonzero, N={len(p_old_nz)})", density=True)
        ax.hist(p_new_nz, bins=bins, alpha=0.5, label=f"NEW (nonzero, N={len(p_new_nz)})", density=True)
        ax.set_xlabel("rec.hdr.pot (per readout event, nonzero only)")
        ax.set_ylabel("density")
        ax.set_title("POT per readout event distribution (sanity check)")
        ax.legend()
        plt.tight_layout()
        pdf.savefig(fig)
    plt.close(fig)

    pdf.close()
    print(f"\nPlots saved to {output_pdf}")


def main():
    old = analyze_file(OLD_PATH, "OLD")
    new = analyze_file(NEW_PATH, "NEW")
    make_comparison_table(old, new)
    make_plots(old, new, OUTPUT_PDF)


if __name__ == "__main__":
    main()
```

```python
"""
Profile the zero-interaction events in the NEW sample.

These are events where ndlp_true == 0 AND ndlp == 0, despite ngenevt == 20.
Something in SPINE is dropping all interactions for ~7.5% of readout events.

Questions this script answers:
  1. Are these events also mc.nnu == 0? (no neutrinos at all vs neutrinos present but SPINE missed them)
  2. Do they correlate with first_in_subrun or first_in_file?
  3. Do they cluster in specific runs/subruns/events?
  4. Do they have unusual POT values?
  5. What does the mc.nu content look like for these events? (energy, pdg, iscc, vertex position)
  6. Are they consecutive events? (suggests a file-level or subrun-level failure)
  7. What fraction of total POT do they carry?
"""

import uproot
import numpy as np

NEW_PATH = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/NewOldStudies/new_file.root"
OLD_PATH = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/NewOldStudies/old_file.root"

# All branches we want to read
BRANCHES = [
    "rec.hdr.pot",
    "rec.hdr.ngenevt",
    "rec.hdr.evt",
    "rec.hdr.first_in_subrun",
    "rec.hdr.first_in_file",
    "rec.hdr.ismc",
    "rec.hdr.husk",
    "rec.mc.nnu",
    "rec.mc.nu..length",
    "rec.ndlp_true",
    "rec.ndlp",
    "rec.dlp_true..length",
    "rec.dlp..length",
]

# MC neutrino branches — jagged, need special handling
MC_NU_BRANCHES = [
    "rec.mc.nu.E",
    "rec.mc.nu.pdg",
    "rec.mc.nu.iscc",
    "rec.mc.nu.isvtxcont",
    "rec.mc.nu.position.x",
    "rec.mc.nu.position.y",
    "rec.mc.nu.position.z",
    "rec.mc.nu.genie_mode",
]


def profile_zeros(path, label):
    print(f"\n{'='*80}")
    print(f"ZERO-INTERACTION EVENT PROFILE: {label}")
    print(f"{'='*80}")

    f = uproot.open(path)
    tree = f["recTree"]
    n_total = tree.num_entries

    # Read scalar branches
    available = set(tree.keys())
    branches_to_read = [b for b in BRANCHES if b in available]
    missing = [b for b in BRANCHES if b not in available]
    if missing:
        print(f"\n  Missing branches (skipped): {missing}")

    arrs = tree.arrays(branches_to_read, library="np")

    ndlp_true = arrs["rec.ndlp_true"]
    ndlp      = arrs["rec.ndlp"]
    mc_nnu    = arrs["rec.mc.nnu"]
    ngenevt   = arrs["rec.hdr.ngenevt"]
    pot       = arrs["rec.hdr.pot"]
    evt       = arrs.get("rec.hdr.evt", np.zeros(n_total))
    first_sub = arrs.get("rec.hdr.first_in_subrun", np.zeros(n_total))
    first_file = arrs.get("rec.hdr.first_in_file", np.zeros(n_total))
    husk      = arrs.get("rec.hdr.husk", np.zeros(n_total))

    # Mask for zero-interaction events
    zero_mask = ndlp_true == 0
    n_zero = zero_mask.sum()
    n_nonzero = (~zero_mask).sum()

    print(f"\n  Total events:              {n_total}")
    print(f"  Events with ndlp_true==0:  {n_zero}  ({100*n_zero/n_total:.2f}%)")
    print(f"  Events with ndlp_true>0:   {n_nonzero}  ({100*n_nonzero/n_total:.2f}%)")

    if n_zero == 0:
        print("  No zero-interaction events found. Nothing to profile.")
        return

    # ---- Check 1: mc.nnu in zero events ----
    print(f"\n--- CHECK 1: mc.nnu in zero-interaction events ---")
    nnu_zero = mc_nnu[zero_mask]
    nnu_nonzero = mc_nnu[~zero_mask]
    unique_z, counts_z = np.unique(nnu_zero, return_counts=True)
    print(f"  mc.nnu distribution (zero events):")
    for u, c in zip(unique_z, counts_z):
        print(f"    mc.nnu == {u}: {c}  ({100*c/n_zero:.1f}%)")
    print(f"  mc.nnu distribution (nonzero events):")
    unique_nz, counts_nz = np.unique(nnu_nonzero, return_counts=True)
    for u, c in zip(unique_nz[:6], counts_nz[:6]):
        print(f"    mc.nnu == {u}: {c}  ({100*c/n_nonzero:.1f}%)")

    # ---- Check 2: ngenevt in zero events ----
    print(f"\n--- CHECK 2: ngenevt in zero-interaction events ---")
    ngen_zero = ngenevt[zero_mask]
    unique_g, counts_g = np.unique(ngen_zero, return_counts=True)
    print(f"  ngenevt distribution (zero events):")
    for u, c in zip(unique_g, counts_g):
        print(f"    ngenevt == {u}: {c}  ({100*c/n_zero:.1f}%)")

    # ---- Check 3: first_in_subrun correlation ----
    print(f"\n--- CHECK 3: first_in_subrun / first_in_file correlation ---")
    fis_zero = first_sub[zero_mask].astype(bool)
    fis_all  = first_sub.astype(bool)
    print(f"  first_in_subrun among zero events: {fis_zero.sum()}/{n_zero} ({100*fis_zero.mean():.1f}%)")
    print(f"  first_in_subrun overall:           {fis_all.sum()}/{n_total} ({100*fis_all.mean():.1f}%)")
    if "rec.hdr.first_in_file" in arrs:
        fif_zero = first_file[zero_mask].astype(bool)
        fif_all  = first_file.astype(bool)
        print(f"  first_in_file among zero events:   {fif_zero.sum()}/{n_zero} ({100*fif_zero.mean():.1f}%)")
        print(f"  first_in_file overall:             {fif_all.sum()}/{n_total} ({100*fif_all.mean():.1f}%)")

    # ---- Check 4: husk flag ----
    print(f"\n--- CHECK 4: husk flag ---")
    husk_zero = husk[zero_mask].astype(bool)
    husk_all  = husk.astype(bool)
    print(f"  husk among zero events: {husk_zero.sum()}/{n_zero} ({100*husk_zero.mean():.1f}%)")
    print(f"  husk overall:           {husk_all.sum()}/{n_total} ({100*husk_all.mean():.1f}%)")

    # ---- Check 5: POT carried by zero events ----
    print(f"\n--- CHECK 5: POT in zero-interaction events ---")
    pot_zero = pot[zero_mask]
    pot_total = pot.sum()
    pot_zero_sum = pot_zero.sum()
    print(f"  POT in zero events:    {pot_zero_sum:.4e}  ({100*pot_zero_sum/pot_total:.2f}% of total)")
    print(f"  POT in nonzero events: {(pot_total - pot_zero_sum):.4e}")

    # ---- Check 6: Are zero events consecutive? ----
    print(f"\n--- CHECK 6: Are zero events consecutive (clustered)? ---")
    zero_indices = np.where(zero_mask)[0]
    diffs = np.diff(zero_indices)
    n_consecutive = (diffs == 1).sum()
    print(f"  Zero-event indices: first={zero_indices[0]}, last={zero_indices[-1]}")
    print(f"  Consecutive pairs:  {n_consecutive}/{len(diffs)}")
    if len(diffs) > 0:
        print(f"  Gap distribution:   min={diffs.min()}, max={diffs.max()}, mean={diffs.mean():.1f}, median={np.median(diffs):.0f}")

    # Check if they come in blocks of 20 (one per subrun)
    if n_zero >= 20:
        block_sizes = []
        current_block = 1
        for d in diffs:
            if d == 1:
                current_block += 1
            else:
                block_sizes.append(current_block)
                current_block = 1
        block_sizes.append(current_block)
        block_sizes = np.array(block_sizes)
        print(f"  Block sizes (consecutive runs): count={len(block_sizes)}, "
              f"mean={block_sizes.mean():.1f}, min={block_sizes.min()}, max={block_sizes.max()}")
        unique_bs, counts_bs = np.unique(block_sizes, return_counts=True)
        print(f"  Block size distribution: {dict(zip(unique_bs[:10], counts_bs[:10]))}")

    # ---- Check 7: ndlp also zero? ----
    print(f"\n--- CHECK 7: ndlp (reco) in zero-truth events ---")
    ndlp_in_zero = ndlp[zero_mask]
    unique_r, counts_r = np.unique(ndlp_in_zero, return_counts=True)
    print(f"  ndlp distribution (zero-truth events):")
    for u, c in zip(unique_r[:5], counts_r[:5]):
        print(f"    ndlp == {u}: {c}  ({100*c/n_zero:.1f}%)")

    # ---- Check 8: MC neutrino properties in zero events ----
    print(f"\n--- CHECK 8: MC neutrino properties in zero vs nonzero events ---")
    mc_nu_available = [b for b in MC_NU_BRANCHES if b in available]
    if mc_nu_available:
        mc_arrs = tree.arrays(mc_nu_available, library="ak")
        import awkward as ak

        for bname in mc_nu_available:
            short = bname.split(".")[-1]
            arr = mc_arrs[bname]
            vals_zero = ak.flatten(arr[zero_mask])
            vals_nonzero = ak.flatten(arr[~zero_mask])

            if len(vals_zero) == 0:
                print(f"  {short:15s}  zero events: NO neutrinos at all")
                continue

            vz = ak.to_numpy(vals_zero)
            vnz = ak.to_numpy(vals_nonzero)

            if short in ("E", "x", "y", "z"):
                print(f"  {short:15s}  zero: mean={vz.mean():.3f}  std={vz.std():.3f}  "
                      f"nonzero: mean={vnz.mean():.3f}  std={vnz.std():.3f}")
            elif short in ("pdg", "iscc", "genie_mode", "isvtxcont"):
                unique_vz, counts_vz = np.unique(vz.astype(int), return_counts=True)
                unique_vnz, counts_vnz = np.unique(vnz.astype(int), return_counts=True)
                dist_z = ", ".join(f"{u}:{c}" for u, c in zip(unique_vz[:6], counts_vz[:6]))
                dist_nz = ", ".join(f"{u}:{c}" for u, c in zip(unique_vnz[:6], counts_vnz[:6]))
                print(f"  {short:15s}  zero: {dist_z}")
                print(f"  {'':15s}  nonzero: {dist_nz}")
    else:
        print("  No mc.nu branches available for this check.")

    # ---- Summary ----
    print(f"\n{'='*80}")
    print("SUMMARY")
    print(f"{'='*80}")
    print(f"  {n_zero} events ({100*n_zero/n_total:.1f}%) have ndlp_true == 0 in the NEW sample.")
    if n_zero > 0 and (nnu_zero > 0).all():
        print(f"  ALL of these events have mc.nnu > 0 (neutrinos ARE present in MCTruth).")
        print(f"  → SPINE is failing to construct ANY truth interactions despite neutrinos existing.")
    elif n_zero > 0 and (nnu_zero == 0).all():
        print(f"  ALL of these events have mc.nnu == 0 (no neutrinos in MCTruth).")
        print(f"  → These are genuinely empty events; SPINE is correct to find nothing.")
    else:
        n_has_nu = (nnu_zero > 0).sum()
        n_no_nu = (nnu_zero == 0).sum()
        print(f"  {n_has_nu} of these events have mc.nnu > 0 (SPINE failed).")
        print(f"  {n_no_nu} of these events have mc.nnu == 0 (genuinely empty).")


# Run on both files
for path, label in [(OLD_PATH, "OLD"), (NEW_PATH, "NEW")]:
    profile_zeros(path, label)
```
