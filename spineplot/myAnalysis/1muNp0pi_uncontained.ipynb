```python
import uproot, numpy as np
f = uproot.open("/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/1muNp0pi_Nge1_uncontained_purity.root")
trees = ["cut1_fiducial", "cut2_flash", "cut3_no_photons", "cut4_no_electrons",
         "cut5_single_muon", "cut6_no_charged_pions", "cut7_has_protons",
         "cut8_trigger_emulation", "cut9_full_selection"]

print(f"{'Stage':<25} {'N_total':>10} {'N_signal':>10} {'Purity':>10}")
for tn in trees:
    cats = f[f"events/nominal/{tn}"]["true_category"].array(library="np")
    n_total = len(cats)
    n_signal = np.isin(cats, [0, 1, 3, 4]).sum()
    purity = n_signal / n_total if n_total > 0 else 0
    print(f"{tn:<25} {n_total:>10} {n_signal:>10} {purity:>10.4f}")
```
